```python
import pandas as pd 
import matplotlib.pyplot as plt
df=pd.read_csv("linkdin_Job_data.csv")
df.head()
```

top 10 location graph

```python
df['location'].value_counts().head(10).plot(kind='bar')

plt.title("Top job locations")
plt.xlabel("locations")
plt.ylabel("number of jobs")

plt.xticks(rotation=45)

plt.show()
```

work type pie chart

```python
df['work_type'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title("work type distribution")

plt.ylabel("")

plt.show()
```

most common job roles

```python
df['job'].value_counts().head(10).plot(kind='bar')

plt.title("Most Common Job Roles")
plt.xlabel("Job Roles")
plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```
